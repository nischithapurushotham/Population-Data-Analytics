```python
import pandas as pd

df = pd.read_csv(
    "API_SP.POP.TOTL_DS2_EN_csv_v2_327505.csv",
    skiprows=4
)
```

```python
df = df.dropna(how="all")
```

```python
df = df.melt(
    id_vars=["Country Name", "Country Code", "Indicator Name", "Indicator Code"],
    var_name="Year",
    value_name="Population"
)
```

```python
df["Year"] = pd.to_numeric(df["Year"], errors="coerce")
df = df.dropna(subset=["Year"])
df["Year"] = df["Year"].astype(int)
```

```python
df["Population"] = pd.to_numeric(df["Population"], errors="coerce")
df = df.dropna(subset=["Population"])
```

```python
df["Population"] = df["Population"].astype(float)
```

```python
df.rename(columns={
    "Country Name":"Country"
}, inplace=True)
```

```python
df["Annual Change"] = df.groupby("Country")["Population"].diff()
```

```python
df["Growth Rate"] = (
    df.groupby("Country")["Population"]
    .pct_change()*100
)
```

```python
df["Decade"] = (df["Year"]//10)*10
```

```python
conditions = [
    df["Growth Rate"] < 0,
    df["Growth Rate"] < 1,
    df["Growth Rate"] < 2,
    df["Growth Rate"] >= 2
]

labels = [
    "Negative",
    "Low",
    "Medium",
    "High"
]

df["Growth Category"] = np.select(
    conditions,
    labels,
    default="Unknown"
)
```

```python
df.to_csv(
    "cleaned_population.csv",
    index=False
)
```
